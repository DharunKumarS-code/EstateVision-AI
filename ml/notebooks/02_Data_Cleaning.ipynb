```python
import pandas as pd
```

```python
df = pd.read_csv("../../datasets/Bengaluru_House_Data.csv")
```

```python
df.isnull().sum()
```

```python
# Remove the society column because it has many missing values
df = df.drop("society", axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df["balcony"].value_counts()
df["balcony"].describe()
```

```python
df["balcony"].value_counts()
df["balcony"].describe()
```

```python
# Fill missing balcony values with the median
df["balcony"] = df["balcony"].fillna(df["balcony"].median())
```

```python
df["balcony"].isnull().sum()
```

```python
df = df.dropna(subset=["location", "size", "bath"])
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df["size"].unique()
```

```python
def extract_bhk(size):
    return int(size.split(" ")[0])
```

```python
print(extract_bhk("2 BHK"))
print(extract_bhk("4 Bedroom"))
print(extract_bhk("1 RK"))
```

```python
df["bhk"] = df["size"].apply(extract_bhk)
```
